```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler, LabelEncoder

# Read the source data
titanic_data = pd.read_csv('Titanic-Dataset.csv')

# View the overview of columns, non-null values, and memory types
print("=== Structural Summary ===")
titanic_data.info()

# Preview the initial data entries
print("\n=== Data Preview ===")
print(titanic_data.head(5))
```

```python
# Handle missing ages by inserting the dataset's average (mean) age
titanic_data['Age'] = titanic_data['Age'].fillna(titanic_data['Age'].mean())

# Patch missing Embarked cities with the most frequent value (mode)
most_frequent_port = titanic_data['Embarked'].mode()[0]
titanic_data['Embarked'] = titanic_data['Embarked'].fillna(most_frequent_port)

# Replace empty Cabin cells with a placeholder string 'Missing'
titanic_data['Cabin'] = titanic_data['Cabin'].fillna('Missing')

print("\nVerification of Remaining Empty Rows:")
print(titanic_data.isnull().sum())
```

```python
# Convert text categoricals into numeric labels
encoder_gender = LabelEncoder()
titanic_data['Sex'] = encoder_gender.fit_transform(titanic_data['Sex'])

encoder_port = LabelEncoder()
titanic_data['Embarked'] = encoder_port.fit_transform(titanic_data['Embarked'].astype(str))

print("\nQuick look at transformed categorical values:")
print(titanic_data[['Sex', 'Embarked']].head(4))
```

```python
# Standardize numeric ranges using Z-score scaling
data_scaler = StandardScaler()
titanic_data[['Age', 'Fare']] = data_scaler.fit_transform(titanic_data[['Age', 'Fare']])

print("\nReview of Scaled Columns:")
print(titanic_data[['Age', 'Fare']].head(4))
```

```python
# 1. Plot distributions to identify anomalies
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.boxplot(titanic_data['Age'])
plt.title('Distribution of Normalized Age')

plt.subplot(1, 2, 2)
plt.boxplot(titanic_data['Fare'])
plt.title('Distribution of Normalized Fare')

plt.tight_layout()
plt.savefig('distribution_outliers.png')
plt.close()

# 2. Extract boundaries using the 1.5 * IQR standard rule
# Age Bounds
q1_a, q3_a = titanic_data['Age'].quantile(0.25), titanic_data['Age'].quantile(0.75)
iqr_a = q3_a - q1_a
low_a, high_a = q1_a - (1.5 * iqr_a), q3_a + (1.5 * iqr_a)

# Fare Bounds
q1_f, q3_f = titanic_data['Fare'].quantile(0.25), titanic_data['Fare'].quantile(0.75)
iqr_f = q3_f - q1_f
low_f, high_f = q1_f - (1.5 * iqr_f), q3_f + (1.5 * iqr_f)

# Keep only values within safe boundaries
filtered_titanic = titanic_data[
    (titanic_data['Age'] >= low_a) & (titanic_data['Age'] <= high_a) &
    (titanic_data['Fare'] >= low_f) & (titanic_data['Fare'] <= high_f)
]

print(f"\nBefore cleaning: {titanic_data.shape[0]} rows")
print(f"After outlier extraction: {filtered_titanic.shape[0]} rows")

# Output the curated file
filtered_titanic.to_csv('cleaned_titanic_output.csv', index=False)
```

```python
from sklearn.model_selection import train_test_split

# Isolate features from the target output column
features = filtered_titanic.drop(columns=['Survived', 'PassengerId', 'Name', 'Ticket', 'Cabin'])
target = filtered_titanic['Survived']

# Split data into train/test subsets (using a different random_state for unique slicing)
train_features, test_features, train_target, test_target = train_test_split(
    features, target, test_size=0.20, random_state=101
)

print("Split operation successful.")
```

```python
from sklearn.linear_model import LogisticRegression

# Initialize a completely different algorithm style: Logistic Regression
classifier_model = LogisticRegression(max_iter=1000, random_state=101)

# Teach the classifier model
classifier_model.fit(train_features, train_target)

print("Logistic Regression training loop completed.")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Generate metrics on unseen test features
predictions = classifier_model.predict(test_features)

# Compute performance totals
score = accuracy_score(test_target, predictions)
print(f"=== Final Model Accuracy: {score * 100:.2f}% ===")

print("\n=== Precision and Recall Matrix ===")
print(classification_report(test_target, predictions))

print("=== Confusion Structure ===")
print(confusion_matrix(test_target, predictions))
```

```python
# Extract weights (coefficients) from the Logistic Regression model
weights = classifier_model.coef_[0]
feature_weights = pd.Series(weights, index=features.columns).sort_values(ascending=False)

print("=== Logistic Regression Feature Weights ===")
print(feature_weights)

# Plot the model weights
plt.figure(figsize=(8, 4))
feature_weights.plot(kind='bar', color='lightcoral')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.title('How Different Features Influence Survival Probability')
plt.ylabel('Weight Value (Coefficient)')
plt.xlabel('Features')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('model_coefficients.png')
plt.close()
print("\nWeight chart exported as 'model_coefficients.png'!")
```

```python
import joblib

# Save the trained Logistic Regression model
joblib.dump(classifier_model, 'titanic_logistic_regression.pkl')

print("Model binary saved cleanly as 'titanic_logistic_regression.pkl'!")
```
